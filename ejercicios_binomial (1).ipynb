# Ejercicios de Distribución Binomial

## Marco teórico

Si $X \sim B(n, p)$:

**PMF:** $P(X=k) = \binom{n}{k} p^k (1-p)^{n-k}$

**CDF:** $F(x) = P(X \le x) = \sum_{k=0}^{x} \binom{n}{k} p^k (1-p)^{n-k}$

**Esperanza:** $E(X) = np$

**Varianza:** $V(X) = np(1-p)$

**Desviación estándar:** $\sigma = \sqrt{np(1-p)}$

```r
# Función para graficar PMF y CDF
library(ggplot2)

graf_binom <- function(n, p, titulo){
  x <- 0:n
  df <- data.frame(x = x, PMF = dbinom(x, n, p), CDF = pbinom(x, n, p))
  
  g1 <- ggplot(df, aes(x = factor(x), y = PMF)) +
    geom_col(fill = 'steelblue', color = 'black') +
    geom_text(aes(label = round(PMF, 3)), vjust = -0.4, size = 3) +
    labs(title = paste('PMF -', titulo), x = 'k', y = 'P(X = k)') +
    theme_minimal()
  
  g2 <- ggplot(df, aes(x = x, y = CDF)) +
    geom_step(color = 'darkred', linewidth = 1) +
    geom_point(color = 'darkred', size = 2) +
    labs(title = paste('CDF -', titulo), x = 'x', y = 'F(x)') +
    scale_x_continuous(breaks = x) +
    theme_minimal()
  
  print(g1); print(g2)
}
```

# Ejercicio 1: Terneros con bajo peso

$X \sim B(15, 0.25)$

```r
n <- 15; p <- 0.25

# a) P(X >= 5)
pa <- 1 - pbinom(4, n, p)
cat('a) P(X >= 5) =', round(pa, 4), '\n')

# b) P(X <= 3)
pb <- pbinom(3, n, p)
cat('b) P(X <= 3) =', round(pb, 4), '\n')

# c) P(X = 8)
pc <- dbinom(8, n, p)
cat('c) P(X = 8) =', round(pc, 6), '\n')
```

**Interpretación c):** Proporción observada 8/15 ≈ 0.533, muy superior al 25%. Probabilidad bajo p=0.25 es muy baja → **el 25% NO se aplica a esta finca**.

```r
# Esperanza, varianza, desviación
mu <- n*p; v <- n*p*(1-p); s <- sqrt(v)
cat('E(X) =', mu, '| V(X) =', v, '| sigma =', round(s,4), '\n')

graf_binom(15, 0.25, 'B(15, 0.25)')
```

# Ejercicio 2: Germinación de semillas

$X \sim B(6, 0.70)$

```r
n <- 6; p <- 0.70

# a) Todas germinen: P(X = 6)
pa <- dbinom(6, n, p)
cat('a) P(X = 6) =', round(pa, 4), '\n')

# b) Menos de 2 NO germinen <=> al menos 5 germinen: P(X >= 5)
pb <- 1 - pbinom(4, n, p)
cat('b) P(X >= 5) =', round(pb, 4), '\n')

# c) Más de 4 germinen: P(X > 4) = P(X >= 5)
pc <- 1 - pbinom(4, n, p)
cat('c) P(X > 4) =', round(pc, 4), '\n')

mu <- n*p; v <- n*p*(1-p); s <- sqrt(v)
cat('\nE(X) =', mu, '| V(X) =', v, '| sigma =', round(s,4), '\n')

graf_binom(6, 0.70, 'B(6, 0.70)')
```

**Interpretación:** En promedio germinan 4.2 de 6 semillas.

# Ejercicio 3: Huevos con defectos

$X \sim B(20, 0.10)$

```r
n <- 20; p <- 0.10

# a) P(X >= 3)
pa <- 1 - pbinom(2, n, p)
cat('a) P(X >= 3) =', round(pa, 4), '\n')

# b) P(X <= 2)
pb <- pbinom(2, n, p)
cat('b) P(X <= 2) =', round(pb, 4), '\n')

# c) P(X = 7)
pc <- dbinom(7, n, p)
cat('c) P(X = 7) =', round(pc, 6), '\n')

mu <- n*p; v <- n*p*(1-p); s <- sqrt(v)
cat('\nE(X) =', mu, '| V(X) =', v, '| sigma =', round(s,4), '\n')

graf_binom(20, 0.10, 'B(20, 0.10)')
```

**Interpretación c):** Proporción 7/20 = 0.35 (triple del 10%). Probabilidad casi nula → **el 10% NO es válido para este lote**.

# Ejercicio 4: Fungicida

X = plantas afectadas, $X \sim B(8, 0.15)$

```r
n <- 8; p <- 0.15

# a) Ninguna afectada: P(X = 0)
pa <- dbinom(0, n, p)
cat('a) P(X = 0) =', round(pa, 4), '\n')

# b) Menos de 2 afectadas: P(X < 2) = P(X <= 1)
pb <- pbinom(1, n, p)
cat('b) P(X < 2) =', round(pb, 4), '\n')

# c) Más de 6 protegidas = menos de 2 afectadas
pc <- pbinom(1, n, p)
cat('c) P(protegidas > 6) =', round(pc, 4), '\n')

mu <- n*p; v <- n*p*(1-p); s <- sqrt(v)
cat('\nE(X) =', mu, '| V(X) =', v, '| sigma =', round(s,4), '\n')

graf_binom(8, 0.15, 'B(8, 0.15) - Afectadas')
```

**Interpretación:** Se esperan 1.2 plantas afectadas de cada 8 tratadas.

# Ejercicio 5: Inseminación artificial

$X \sim B(12, 0.65)$

```r
n <- 12; p <- 0.65

# a) P(X >= 8)
pa <- 1 - pbinom(7, n, p)
cat('a) P(X >= 8) =', round(pa, 4), '\n')

# b) P(X <= 5)
pb <- pbinom(5, n, p)
cat('b) P(X <= 5) =', round(pb, 4), '\n')

# c) P(X = 4)
pc <- dbinom(4, n, p)
cat('c) P(X = 4) =', round(pc, 4), '\n')

mu <- n*p; v <- n*p*(1-p); s <- sqrt(v)
cat('\nE(X) =', mu, '| V(X) =', v, '| sigma =', round(s,4), '\n')

graf_binom(12, 0.65, 'B(12, 0.65)')
```

**Interpretación c):** Proporción 4/12 ≈ 0.33, muy inferior al 65%. Probabilidad baja bajo p=0.65 → **el 65% NO se aplica a este hato** (problemas de manejo, semen o detección de celo).
